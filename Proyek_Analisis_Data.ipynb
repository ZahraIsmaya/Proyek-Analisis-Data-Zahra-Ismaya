### Explore Kolom Waktu

```python
# gabungkan kolom waktu
all_df["datetime"] = pd.to_datetime(
    all_df[["year", "month", "day", "hour"]]
)

# urutkan berdasarkan waktu
all_df = all_df.sort_values(by="datetime")

all_df.head()
```

**Insight:**
- Penggabungan kolom year, month, day, dan hour menjadi satu kolom datetime mengubah data dari menjadi objek waktu yang bermakna untuk analisis time series
- Sorting datetime berguna dalam visualisasi tren agar berurutan

### Explore Feature

```python
# ekstrak bulan & jam
all_df["month"] = all_df["datetime"].dt.month
all_df["hour"] = all_df["datetime"].dt.hour

# kategori kualitas udara (berdasarkan nilai PM2.5)
all_df["air_quality_category"] = pd.cut(
    all_df["PM2.5"],
    bins=[0, 50, 100, 150, all_df["PM2.5"].max()],
    labels=["Baik", "Sedang", "Tidak Sehat", "Berbahaya"]
)

all_df.head()
```

**Insight:**
- Ekstraksi kolom month dan hour untuk melihat siklus harian
- Pembuatan kolom air_quality_category sebagai target dalam melakukan klasifikasi

### Explore Distribusi Data



```python
print("Mean: ", all_df["PM2.5"].mean())
print("Median: ", all_df["PM2.5"].median())
print("Minimum: ", all_df["PM2.5"].min())
```

**Insight:**
- Nilai Mean (79.8) yang jauh lebih besar dari Median (55.0) menunjukkan bahwa distribusi datamiring ke kanan (positively skewed)

- Karena Median (55.0) berada di atas batas kategori "Baik", bisa disimpulkan bahwa secara umum kualitas udara lebih sering berada pada level "Sedang" hingga "Tidak Sehat"

## Visualization & Explanatory Analysis

### Pertanyaan 1: Bagaimana kondisi kualitas udara dalam beberapa waktu terakhir, dan apakah terdapat pola tertentu dari waktu ke waktu?

```python
monthly_trend = all_df.groupby("month")["PM2.5"].mean()

plt.figure()
monthly_trend.plot(marker='o')
plt.title("Rata-rata Kualitas Udara per Bulan")
plt.xlabel("Bulan")
plt.ylabel("Rata-rata Polusi Udara")
plt.show()
```

### Pertanyaan 2: Faktor apa saja yang mempengaruhi baik atau buruknya kualitas udara?

```python
plt.figure()
sns.heatmap(
    all_df[["PM2.5", "TEMP", "WSPM", "PRES", "RAIN"]].corr(),
    annot=True
)
plt.title("Korelasi Antar Variabel")
plt.show()
```

### Pertanyaan 3: Seberapa sering kualitas udara berada dalam kondisi baik, sedang, atau buruk dalam periode tertentu?


```python
all_df["air_quality_category"] = pd.cut(
    all_df["PM2.5"],
    bins=[0, 50, 100, 150, all_df["PM2.5"].max()],
    labels=["Baik", "Sedang", "Tidak Sehat", "Berbahaya"]
)

plt.figure()
sns.countplot(data=all_df, x="air_quality_category")
plt.title("Distribusi Kategori Kualitas Udara")
plt.show()
```

### Pertanyaan 4: Bagaimana perbedaan kondisi kualitas udara di setiap lokasi pengamatan, dan lokasi mana yang paling tercemar?


```python
plt.figure()
sns.barplot(
    data=all_df,
    x="station",
    y="PM2.5"
)
plt.title("Rata-rata Kualitas Udara per Lokasi")
plt.xticks(rotation=45)
plt.show()
```

### Pertanyaan 5: Pada waktu tertentu dalam sehari, kapan kualitas udara cenderung paling buruk?


```python
hourly_trend = all_df.groupby("hour")["PM2.5"].mean()

plt.figure()
hourly_trend.plot(marker='o')
plt.title("Kualitas Udara Berdasarkan Jam")
plt.xlabel("Jam")
plt.ylabel("Rata-rata PM2.5")
plt.show()
```

### Pertanyaan 6: Apakah terdapat kondisi kualitas udara yang sangat buruk (ekstrem), dan seberapa sering hal tersebut terjadi?


```python
plt.figure()
sns.boxplot(x=all_df["PM2.5"])
plt.title("Distribusi Kualitas Udara (Deteksi Ekstrem)")
plt.show()
```

**Insight:**
- Rata-rata Kualitas Udara per Bulan = PM2.5 tinggi di awal tahun (bulan 1-3), turun ke titik terendah di bulan 8, lalu naik kembali di bulan 10-12. Terdapat pola fluktuasi yang konsisten sepanjang tahun..

- Korelasi Antar Variabel = Kecepatan angin (WSPM) paling berpengaruh terhadap PM2.5 (-0.27), diikuti suhu (TEMP) dengan korelasi lemah (-0.13). Curah hujan (RAIN) dan tekanan udara (TEMP) hampir tidak berpengaruh (-0.014 & -0.013).

- Distribusi Kategori Kualitas Udara = Kategori "Baik" paling dominan , diikuti kategori "Sedang" , kemudian kategori "Berbahaya" dan yang paling jarang terjadi kategori "Tidak Sehat".

- Rata-rata Kualitas Udara per Lokasi = Dingling adalah lokasi dengan PM2.5 terendah, sementara Zhanguan dan Shunyi tertinggi. Perbedaan antar lokasi ada namun tidak terpaut sangat jauh.

- Kualitas Udara Berdasarkan Jam = PM2.5 paling tinggi pada jam 22-23 malam, dan paling rendah pada jam 07-08 pagi.

- Distribusi Kualitas Udara (Deteksi Ekstrem) = Mayoritas data terkonsentrasi di rentang rendah, namun terdapat banyak outlier

## Analisis Lanjutan (Opsional)

```python
# PERBANDINGAN MUSIM
season_map = {
    12: "Winter", 1: "Winter", 2: "Winter",
    3: "Spring", 4: "Spring", 5: "Spring",
    6: "Summer", 7: "Summer", 8: "Summer",
    9: "Autumn", 10: "Autumn", 11: "Autumn"
}

all_df["season"] = all_df["month"].map(season_map)
sns.boxplot(data=all_df, x="season", y="PM2.5")
plt.title("Kualitas Udara Berdasarkan Musim")
plt.show()
```

**Insight:**
- Summer memiliki median PM2.5 paling rendah dan box paling kecil sehingga udara pada summer paling bersih dan paling stabil/konsisten
- Winter memiliki median tertinggi, box paling lebar, dan outlier paling banyak sehingga udara pada winter paling buruk sekaligus paling tidak stabil
- Autumn memiliki median sedikit lebih tinggi dari Spring
- Spring berada di tengah, dengan beberapa outlier ekstrem

## Conclusion

- Terdapat pola fluktuasi polusi udara (PM2.5) yang konsisten sepanjang tahun. Polusi tertinggi terjadi di awal dan akhir tahun, sementara pertengahan tahun (sekitar bulan 8) menjadi periode dengan udara paling bersih. Pola ini berulang dan dapat diprediksi.
- Kecepatan angin (WSPM) adalah faktor yang paling berpengaruh, semakin kencang angin, semakin rendah PM2.5. Suhu juga sedikit berpengaruh namun lemah. Sedangkan curah hujan tidak terbukti signifikan memengaruhi kualitas udara.
- Kondisi "Baik" paling sering terjadi, namun yang mengkhawatirkan adalah kondisi "Berbahaya" lebih sering muncul dibanding "Tidak Sehat", menandakan ketika kualitas udara memburuk, eskalasi ke level berbahaya terjadi cukup cepat tanpa melewati fase "Tidak Sehat" terlebih dahulu.
- Dingling adalah lokasi paling bersih dengan rata-rata PM2.5 terendah, sedangkan Zhanguan dan Shunyi adalah lokasi paling tercemar. Meski begitu, perbedaan antar lokasi tidak terlalu ekstrem, menunjukkan polusi menyebar secara merata di seluruh area pengamatan.
- Kualitas udara paling buruk terjadi pada malam hingga dini hari (jam 21-01), dan paling baik pada pagi hari sekitar jam 07-08.
- Kondisi ekstrem terbukti ada, ditandai dengan banyaknya outlier. Frekuensinya cukup signifikan dan tidak bisa diabaikan, sehingga perlu menjadi perhatian serius dalam pengambilan kebijakan pengendalian polusi udara.
- Musim sangat berpengaruh terhadap kualitas udara. Winter (musim dingin) adalah periode paling berbahaya dengan tingkat PM2.5 tertinggi dan kejadian ekstrem paling sering, sedangkan Summer (musim panas) adalah periode paling aman dengan udara paling bersih dan konsisten. Hal ini juga menjawab dan menguatkan temuan pertanyaan 1 sebelumnya  bahwa lonjakan PM2.5 di bulan 10-12 dan 1-3 memang berkaitan dengan musim gugur dan musim dingin, bukan sekadar pola numerik bulanan.
